```python
# counting no. of alphabets,no.s and special characters/symbols
s=input("Enter the string:")
c,k,j=0,0,0
print(len(s))
for i in s:
    if i.isalpha():
        c+=1
    elif i.isalnum():
        k+=1
    else:
        j+=1
print(f" The no. of alphabets in string {s} is {c}")
print(f" The no. of numbers in string {s} is {k}")
print(f" The no. of symbols in string {s} is {j}")
```

```python
# Counting no. of vowels and consonants in a string
s=input("Enter the string:")
c,j=0,0
print(len(s))
for i in s:
    if i in "aeiou":
        c+=1
    elif i not in "aeiou" and i.isalpha():
        j+=1
print(f" The no. of vowels in string {s} is {c}")
print(f" The no. of consonants in string {s} is {j}")
```

```python
# reversing a string without using slicing 
s = input("Enter the string:")
j = ""
for i in range(len(s) - 1, -1, -1):
    j += s[i]
print(f"The reverse of the string {s} without slicing is {j}")
# reversing a string with using slicing 
print(f"The reverse of the string {s} with slicing is {s[::-1]}")
```

```python
#check whether a string is palindrome or not
s = input("Enter the string:")
j = ""
for i in range(len(s) - 1, -1, -1):
    j += s[i]
if j==s:
    print(f"The string {s} is palindrome")
else:
    print("the string",s," is not palindrome")
#with slicing
k = s[::-1]
if k==s:
    print(f"The string {s} is palindrome")
else:
    print(f"the string {s} is not palindrome") 
```

```python
# Counting the frequency of given character 
char=input("Enter a character")
s=input("enter the string")
if len(char) != 1:
    print("Please enter only one character")
c=0
for i in s:
    if i==char.lower()or i==char.upper():#if i.lower() == char.lower():
        c+=1
if c!=0:
    print(f"The frequency of a given character {char} in string {s} is {c}")
else:
    print("The character is not present in the string")
# with method
print(f"The frequency of a given character {char} in string {s} is {s.count(char)}")

```
